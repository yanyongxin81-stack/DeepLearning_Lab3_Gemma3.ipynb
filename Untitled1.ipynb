```python

# ======================================
# Lab 3 - 原始 Gemma 模型生成测试
# ======================================

import torch

prompt = """
请创作一个发生在午夜废弃医院的悬疑故事。

要求：
1. 故事具有强烈的悬疑氛围。
2. 设置至少三个隐藏线索。
3. 故事情节具有逻辑性。
4. 结尾设置一个意想不到的反转。
5. 使用中文，字数约500字。
"""

messages = [
    {
        "role": "user",
        "content": [
            {"type": "text", "text": prompt}
        ]
    }
]

# 将提示词转换为模型输入
inputs = processor.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt"
)

# 将输入放到模型所在设备
inputs = inputs.to(device)

input_length = inputs["input_ids"].shape[-1]

# 固定随机种子，便于后续实验对比
torch.manual_seed(42)

# 开始生成
with torch.inference_mode():
    outputs = model.generate(
        **inputs,
        max_new_tokens=512,
        do_sample=True,
        temperature=0.8,
        top_p=0.9,
        repetition_penalty=1.1
    )

# 提取模型新生成的内容
generated_ids = outputs[0][input_length:]

result = processor.decode(
    generated_ids,
    skip_special_tokens=True
)

print("========== 原始 Gemma 生成结果 ==========")
print(result)
```

```python

# ==========================================
# Lab 3 - 实验一：原始 Gemma 模型生成测试
# ==========================================

import torch

# 确保模型处于推理模式
model.eval()

# 1. 设计悬疑故事提示词
prompt = """
请创作一个发生在午夜废弃医院的悬疑故事。

具体要求：
1. 故事具有强烈的悬疑和恐怖氛围。
2. 设置至少三个隐藏线索。
3. 故事情节具有逻辑性。
4. 结尾设置一个意想不到的反转。
5. 使用中文，篇幅约300字。
"""

# 2. 构建对话格式
messages = [
    {
        "role": "user",
        "content": [
            {"type": "text", "text": prompt}
        ]
    }
]

# 3. 将文本转换成模型输入
inputs = processor.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt"
)

# 4. 检查输入 Token 是否正常
input_ids = inputs["input_ids"]
vocab_size = model.get_input_embeddings().num_embeddings

print("输入 Token 数量：", input_ids.shape[-1])
print("Token 最大编号：", input_ids.max().item())
print("模型词表大小：", vocab_size)

assert input_ids.min().item() >= 0
assert input_ids.max().item() < vocab_size

# 5. 将输入移动到模型所在设备
input_device = model.get_input_embeddings().weight.device

inputs = {
    k: v.to(input_device) if isinstance(v, torch.Tensor) else v
    for k, v in inputs.items()
}

input_length = inputs["input_ids"].shape[-1]

# 6. 执行文本生成
print("\n正在生成悬疑故事，请稍等...")

with torch.inference_mode():
    outputs = model.generate(
        **inputs,
        max_new_tokens=256,
        do_sample=True,
        temperature=0.8,
        top_p=0.9,
        repetition_penalty=1.1
    )

# 7. 提取新生成的文本
generated_ids = outputs[0][input_length:]

result = processor.tokenizer.decode(
    generated_ids,
    skip_special_tokens=True
)

print("\n========== 原始 Gemma 生成结果 ==========\n")
print(result)

# 8. 保存生成结果
with open(
    "/content/baseline_result.txt",
    "w",
    encoding="utf-8"
) as f:
    f.write("实验三：原始 Gemma 生成结果\n\n")
    f.write("提示词：\n")
    f.write(prompt)
    f.write("\n\n生成结果：\n")
    f.write(result)

print("\n原始模型生成结果已保存！")
```

```python

# ==========================================
# Lab 3 - Step 5：构建中文悬疑故事数据集
# 使用 Gemma 3 4B 生成合成训练样本
# ==========================================

import torch
import json
import random
import os

# 检查模型是否仍然存在
assert model is not None, "请先加载 Gemma 模型"
assert processor is not None, "请先加载 Processor"

model.eval()

# 1. 设置故事场景
scenes = [
    "午夜的废弃医院",
    "暴雨中的山间旅馆",
    "深夜的老火车站",
    "一座被封锁的古老博物馆",
    "没有住户的公寓大楼",
    "海边的废弃灯塔",
    "深夜的地下停车场",
    "小镇上的旧图书馆",
    "浓雾中的夜间渡轮",
    "已经关闭多年的旧学校"
]

# 2. 设置写作任务
tasks = [
    "创作一篇完整的悬疑故事，要求有隐藏线索和反转结局。",
    "采用第一人称叙事，让读者逐渐发现叙述者隐瞒了真相。",
    "设计一个侦探推理故事，至少包含三个有效线索。",
    "设计一个密室案件，最终给出合理的破解过程。",
    "通过两个人物之间的对话，逐渐揭示隐藏的秘密。",
    "创作一个具有强烈悬疑感的故事开头。",
    "围绕一个不起眼的物品设计一个意外反转。",
    "采用日记或书信形式讲述一件神秘事件。",
    "设计三个嫌疑人，利用不同的证词揭示真相。",
    "创作一个开放式悬疑结尾，让读者留下疑问。"
]

# 3. 构建100条不同的指令
prompts = []

for scene in scenes:
    for task in tasks:
        prompt = f"""
请用中文完成以下悬疑写作任务。

故事场景：{scene}

写作要求：{task}

其他要求：
1. 情节合理，前后连贯。
2. 注意悬疑氛围的营造。
3. 线索应当与故事结局有关。
4. 不要直接解释所有悬念。
5. 输出完整、自然的中文内容。
"""
        prompts.append(prompt)

print("待生成数据数量：", len(prompts))

# 4. 逐条生成训练样本
records = []

input_device = model.get_input_embeddings().weight.device

for i, prompt in enumerate(prompts):

    messages = [
        {
            "role": "user",
            "content": [
                {"type": "text", "text": prompt}
            ]
        }
    ]

    inputs = processor.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt"
    )

    inputs = {
        k: v.to(input_device) if isinstance(v, torch.Tensor) else v
        for k, v in inputs.items()
    }

    input_length = inputs["input_ids"].shape[-1]

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=160,
            do_sample=False,
            use_cache=True
        )

    generated_ids = outputs[0][input_length:]

    completion = processor.tokenizer.decode(
        generated_ids,
        skip_special_tokens=True
    ).strip()

    if len(completion) < 30:
        print(f"第 {i+1} 条生成内容过短，暂时跳过")
        continue

    # 构建完整对话格式
    conversation = [
        {"role": "user", "content": prompt},
        {"role": "assistant", "content": completion}
    ]

    formatted_text = processor.apply_chat_template(
        conversation,
        tokenize=False,
        add_generation_prompt=False
    )

    records.append({
        "prompt": prompt,
        "completion": completion,
        "text": formatted_text
    })

    if (i + 1) % 10 == 0:
        print(
            f"已处理：{i+1}/100，"
            f"有效样本：{len(records)}"
        )

# 5. 随机打乱数据
random.Random(42).shuffle(records)

# 6. 划分训练集、验证集和测试集
n = len(records)

train_end = int(n * 0.8)
val_end = int(n * 0.9)

train_data = records[:train_end]
val_data = records[train_end:val_end]
test_data = records[val_end:]

# 7. 保存 JSONL 文件
DATA_DIR = "/content/Lab3_Dataset"
os.makedirs(DATA_DIR, exist_ok=True)

def save_jsonl(data, filename):
    path = os.path.join(DATA_DIR, filename)

    with open(path, "w", encoding="utf-8") as f:
        for item in data:
            f.write(json.dumps(item, ensure_ascii=False) + "\n")

    return path

save_jsonl(train_data, "train.jsonl")
save_jsonl(val_data, "validation.jsonl")
save_jsonl(test_data, "test.jsonl")
save_jsonl(records, "all_data.jsonl")

# 8. 输出统计信息
print("\n========== 数据集构建完成 ==========")
print("有效样本总数：", n)
print("训练集：", len(train_data))
print("验证集：", len(val_data))
print("测试集：", len(test_data))
print("保存目录：", DATA_DIR)

# 9. 展示前两个样本
for i, item in enumerate(records[:2]):
    print(f"\n========== 样本 {i+1} ==========")
    print("指令：", item["prompt"])
    print("回答：", item["completion"])

print("\n数据集生成任务完成！")
```
